```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
import joblib

# 1. Load the dataset
df = pd.read_csv('../data/raw/cleaned_credit_risk_dataset.csv')

# 2. Identify raw categorical columns present in the dataset
target_cats = ['Employment_Type', 'Loan_Purpose']
cat_cols = [col for col in target_cats if col in df.columns]

if cat_cols:
    encoder = OneHotEncoder(sparse_output=False, handle_unknown='ignore')
    encoded_data = encoder.fit_transform(df[cat_cols])
    joblib.dump(encoder, '../models/encoder.pkl')
    print(f"Encoder fitted on {cat_cols} and saved to ../models/encoder.pkl")
else:
    print("Dataset is already pre-encoded. No raw string columns found.")
    print("Run 'python create_encoder.py' in terminal if you need to build encoder.pkl separately.")
```
